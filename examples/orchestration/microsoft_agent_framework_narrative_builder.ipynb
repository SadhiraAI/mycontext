````python
from mycontext.intelligence import get_pattern_class

NarrativeBuilder = get_pattern_class("narrative_builder")
topic = "Our company's digital transformation journey"

ctx = NarrativeBuilder().build_context(
    topic=topic,
    audience="investors",
    goal="Inspire confidence and showcase progress"
)
instruction = ctx.assemble()
print("Context built. Instruction length:", len(instruction), "chars")
````

````python
try:
    import asyncio

    from agent_framework.openai import OpenAIChatClient
except ImportError:
    raise ImportError("pip install agent-framework --pre")

async def run_agent():
    agent = OpenAIChatClient(model_id="gpt-4o-mini").as_agent(
        name="narrative_builder",
        instructions=instruction,
    )
    result = await agent.run(f"Create a short narrative about: {topic}")
    return result

result = await run_agent()
print(result)
print("\n--- Template works with Microsoft Agent Framework ---")
````
