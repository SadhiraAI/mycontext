````python
from mycontext.intelligence import get_pattern_class
from mycontext.integrations import SemanticKernelHelper

TechnicalTranslator = get_pattern_class("technical_translator")
content = "The API employs idempotent semantics with eventual consistency guarantees."

ctx = TechnicalTranslator().build_context(
    technical_text=content,
    target_audience="non-technical stakeholders"
)
print("Context built. Length:", len(ctx.assemble()), "chars")
````

````python
try:
    from semantic_kernel import Kernel
    from semantic_kernel.connectors.ai.open_ai import OpenAIChatCompletion
except ImportError:
    raise ImportError("pip install semantic-kernel")

kernel = Kernel()
kernel.add_service(OpenAIChatCompletion(ai_model_id="gpt-4o-mini"))

prompt = SemanticKernelHelper.to_prompt_template(ctx)
prompt += f"\n\n**Content to translate:**\n{content}"

result = await kernel.invoke_prompt(prompt)
print(result)
print("\n--- Template works with Semantic Kernel ---")
````
