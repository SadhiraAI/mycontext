````python
from mycontext.intelligence import get_pattern_class

ConceptExplainer = get_pattern_class("concept_explainer")
concept = "quantum entanglement"
audience = "high school students"

ctx = ConceptExplainer().build_context(
    concept=concept,
    audience=audience,
    depth="comprehensive",
    context=""
)
instruction = ctx.assemble()
print("Context built. Instruction length:", len(instruction), "chars")
````

````python
try:
    from mycontext.integrations import GoogleADKHelper
    from google.adk.runners import InMemoryRunner
except ImportError:
    raise ImportError("pip install mycontext-ai google-adk")

agent = GoogleADKHelper.create_agent(
    ctx,
    name="concept_explainer",
    model="gemini-2.0-flash",
    description="Explains complex concepts clearly with layered depth",
)

runner = InMemoryRunner(agent=agent)
result = await runner.run_debug(f"Explain {concept} for {audience}")
print(result)
print("\n--- Template works with Google ADK ---")
````
