# AutoGen (Microsoft) + mycontext RiskAssessor

Simple integration: **RiskAssessor** template as AutoGen assistant system message → verify templates work.

**Template:** `risk_assessor` — likelihood, impact, mitigation strategies

**Prerequisites:**
```bash
pip install "mycontext-ai[openai]" pyautogen
```
Set `OPENAI_API_KEY`.

````python
from mycontext.intelligence import get_pattern_class
from mycontext.integrations import AutoGenHelper

RiskAssessor = get_pattern_class("risk_assessor")
situation = "Migrating customer DB to new cloud provider with 2-week cutover window."

ctx = RiskAssessor().build_context(
    decision=situation,
    context="FinTech, regulatory constraints",
    depth="comprehensive"
)
print("Context built. Length:", len(ctx.assemble()), "chars")
````

````python
try:
    import autogen
except ImportError:
    raise ImportError("pip install pyautogen")

config = {"config_list": [{"model": "gpt-4o-mini", "api_key": "${OPENAI_API_KEY}"}]}
assistant = AutoGenHelper.create_assistant(ctx, name="risk_analyst", llm_config=config)
user_proxy = AutoGenHelper.create_user_proxy(name="user")

user_proxy.initiate_chat(assistant, message=f"Assess risks for: {situation}")
print("\n--- Template works with AutoGen ---")
````
