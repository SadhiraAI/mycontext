````python
from mycontext.integrations import CrewAIHelper
from mycontext.intelligence import get_pattern_class

FeedbackComposer = get_pattern_class("feedback_composer")
work = "Research report on market trends - good structure but weak citations."

ctx = FeedbackComposer().build_context(
    situation=work,
    goal="Improve structure, citations, and clarity"
)
print("Context built. Length:", len(ctx.assemble()), "chars")
````

````python
try:
    from crewai import Agent, Crew, Task
except ImportError:
    raise ImportError("pip install crewai crewai-tools")

agent = CrewAIHelper.create_agent(ctx, name="feedback_expert")
task = CrewAIHelper.create_task(ctx, description=f"Provide feedback on: {work}", agent=agent)

crew = Crew(agents=[agent], tasks=[task])
result = crew.kickoff()
print(str(result)[:100])
print("\n--- Template works with CrewAI ---")
````
