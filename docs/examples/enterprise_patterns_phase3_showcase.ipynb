# Part 1: Temporal Reasoning (3)

**What**: Analyze events across time, plan future scenarios, learn from history.

**Research**: Allen (1983) temporal intervals, Klein (1989) recognition-primed decisions, Schoemaker (1995) scenario planning

**When to use**: Strategic planning, historical analysis, timeline construction, forecasting

## 1. TemporalSequenceAnalyzer

**What it does**: Analyze events across time to identify patterns, trends, and causal sequences.

**When to use**: 
- Historical analysis
- Timeline construction
- Causal chain detection
- Trend forecasting

**Industry examples**:
- **Financial Services**: Analyzing market events leading to 2008 crisis
- **Healthcare**: Understanding disease progression timelines
- **Business Intelligence**: Customer journey temporal analysis
- **Security**: Incident timeline reconstruction

```python
# Example: Business - Product launch timeline analysis
temporal = TemporalSequenceAnalyzer()

# Context-Only Mode (build context without LLM)
context = temporal.build_context(
    events="""Month 1: Product beta launch, 100 early adopters
Month 2: Viral social media post, 10,000 signups
Month 3: Competitor announces similar product
Month 4: Growth slows, churn increases
Month 5: Major feature update
Month 6: Growth recovers, positive reviews""",
    time_span="6 months",
    context="SaaS startup product launch"
)

display(Markdown(f"**Context built** - {len(context.directive.content)} chars\n\n" + context.to_markdown()))
```

```python
# Full LLM Mode (execute with OpenAI)
result = temporal.execute(
    provider="openai",
    events="""Week 1: Database queries suddenly 10x slower
Week 2: Added more database replicas, no improvement
Week 3: Customer complaints escalate
Week 4: DBA discovers missing index on frequently queried column
Week 5: Index added, performance restored""",
    time_span="5 weeks",
    context="Production database performance issue"
)

display(Markdown("### Temporal Analysis Result\n\n" + (result.response[:1500] + "\n\n*...truncated*" if len(result.response) > 1500 else result.response)))
```

```python
# Hybrid Mode (build context, modify, then execute)
context = temporal.build_context(
    events="Economic indicators: inflation rising, unemployment dropping, GDP growth slowing",
    time_span="2022-2024"
)

# Add custom knowledge
from mycontext import Knowledge
context.add_knowledge(
    Knowledge(
        content="Focus on leading vs lagging indicators and predictive signals",
        source="Economic analysis guidelines"
    )
)

# Execute modified context
result = context.execute(provider="openai")
display(Markdown("### Hybrid result with custom knowledge\n\n" + (result.response[:1200] + "\n\n*...truncated*" if len(result.response) > 1200 else result.response)))
```

**When to use**:

- Incident analysis
- Quality improvement
- Process optimization

**Industry examples**:
- **Toyota**: Five Whys methodology originated here
- **Aviation**: Incident investigation and safety analysis
- **Software**: Production incident postmortems
- **Manufacturing**: Quality defect analysis

```python
# Example: Software production incident
root_cause = RootCauseAnalyzer()

result = root_cause.execute(
    provider="openai",
    problem="Website outage - 2 hours downtime",
    symptoms="""- 503 errors for all users
- Load balancer showed all backends unhealthy
- Database connections maxed out
- Started after routine deployment"""
)

display(Markdown("### Root Cause Analysis (Five Whys + Ishikawa)\n\n" + (result.response[:1500] + "\n\n*...truncated*" if len(result.response) > 1500 else result.response)))
```

## 5. DifferentialDiagnoser

**What it does**: Medical-style differential diagnosis for any domain - systematic evaluation of competing hypotheses.

**When to use**:
- Technical troubleshooting (multiple possible causes)
- Business problem diagnosis
- System debugging
- Hypothesis testing

**Industry examples**:
- **Medicine**: Primary application - rule out competing diagnoses
- **IT Support**: Network troubleshooting with multiple possible causes
- **Consulting**: Business problem diagnosis
- **Engineering**: System failure analysis

```python
# Example: Performance issue with multiple hypotheses
differential = DifferentialDiagnoser()

result = differential.execute(
    provider="openai",
    presenting_problem="Mobile app crashes for 15% of Android users",
    observed_data="""- Only affects Android 12+
- Crash occurs during image upload
- Started after last release
- No errors in logs before crash
- Works fine on iOS and older Android""",
    domain="Mobile app development"
)

display(Markdown("### Differential Diagnosis (Ranked hypotheses)\n\n" + (result.response[:1500] + "\n\n*...truncated*" if len(result.response) > 1500 else result.response)))
```

# Comparing Phase 3 Patterns

## When to Use Which Pattern?

### Temporal Patterns
- **TemporalSequenceAnalyzer**: Past events → identify what happened and why
- **FutureScenarioPlanner**: Future planning → explore multiple possible futures
- **HistoricalContextMapper**: Present decisions → learn from historical precedents

### Diagnostic Patterns
- **RootCauseAnalyzer**: Known problem → find underlying cause
- **DifferentialDiagnoser**: Unknown cause → evaluate competing hypotheses
- **SystemHealthAuditor**: No specific problem → comprehensive preventive check

### Synthesis Patterns
- **CrossDomainSynthesizer**: Innovation → borrow insights from other domains
- **PatternRecognitionEngine**: Data/observations → identify recurring patterns
- **HolisticIntegrator**: Multiple perspectives → unified understanding

## Real-World Workflow Example

**Scenario**: Tech company facing competitive threat

1. **HistoricalContextMapper**: Learn from past disruptions (Kodak, Blockbuster)
2. **TemporalSequenceAnalyzer**: Analyze recent market events and trends
3. **FutureScenarioPlanner**: Create 4 scenarios for next 5 years
4. **CrossDomainSynthesizer**: Borrow strategies from other industries
5. **HolisticIntegrator**: Integrate technical, business, customer perspectives
6. **SystemHealthAuditor**: Assess current capabilities and readiness

```python
# Combined workflow example
situation = "New AI competitor disrupting our market"

# Step 1: Learn from history
historical_result = HistoricalContextMapper().execute(
    provider="openai",
    current_situation=situation,
    question="What can we learn from past technology disruptions?"
)

# Step 2: Analyze recent timeline
temporal_result = TemporalSequenceAnalyzer().execute(
    provider="openai",
    events="Month 1: Competitor launches. Month 2: Our market share drops 10%. Month 3: Customer surveys show feature gaps.",
    time_span="3 months"
)

# Step 3: Plan future scenarios
scenario_result = FutureScenarioPlanner().execute(
    provider="openai",
    focal_question="How will AI transform our industry?",
    time_horizon="5 years",
    current_situation=situation
)

print("Comprehensive Strategic Analysis Complete!")
print(f"Historical insights: {len(historical_result.response)} chars")
print(f"Temporal analysis: {len(temporal_result.response)} chars")
print(f"Future scenarios: {len(scenario_result.response)} chars")
```
