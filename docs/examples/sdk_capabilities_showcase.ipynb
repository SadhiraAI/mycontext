## 5. Pattern Suggestion Engine [NO API for keyword, API for LLM/hybrid]

Automatically match templates to questions using keyword analysis, LLM reasoning, or both.

```python
from mycontext.intelligence import suggest_patterns

# Keyword mode (no API call)
result = suggest_patterns(
    "Should we migrate our monolith to microservices? What are the trade-offs?",
    mode="keyword",
    max_patterns=5
)

print(f"Source: {result.source}")
print(f"Suggested: {len(result.suggested_patterns)} patterns")
for s in result.suggested_patterns:
    print(f"  {s.name} [{s.category}] — {s.reason[:80]}")
if result.suggested_chain:
    print(f"\nWorkflow chain: {result.suggested_chain}")
```

```python
# Hybrid mode (keyword + LLM) [API]
result = suggest_patterns(
    "Our AI model is producing biased hiring recommendations. How do we diagnose and fix this?",
    mode="hybrid",
    llm_provider=PROVIDER,
    max_patterns=3
)

print(f"Source: {result.source}")
for s in result.suggested_patterns:
    pos = f" (step {s.chain_position})" if s.chain_position else ""
    print(f"  {s.name}{pos} [{s.category}] conf={s.confidence:.0%}")
    print(f"    {s.reason[:100]}")
```

```python
# Export suggestions in multiple formats
print("=== Markdown ===")
print(result.to_markdown())

print("\n=== Dict (for APIs) ===")
import json

print(json.dumps(result.to_dict(), indent=2)[:500])
```

## 13. Output Evaluation — 5-Dimension Quality Scoring [API]

Evaluate LLM output quality across Instruction Following, Reasoning Depth, Actionability, Structure Compliance, and Cognitive Scaffolding.

```python
from mycontext.intelligence import OutputEvaluator

# Build a context and get a response
ctx = RootCauseAnalyzer().build_context(
    problem="Server crashes during peak hours",
    depth="standard"
)
result = ctx.execute(provider=PROVIDER)

# Evaluate the output
evaluator = OutputEvaluator(mode='llm', provider=PROVIDER)
score = evaluator.evaluate(ctx, result.response)

print(f"Overall: {score.overall:.1%}")
print("\nDimension scores:")
for dim, val in score.dimensions.items():
    print(f"  {dim.value}: {val:.0%}")
print(f"\nStrengths: {score.strengths}")
print(f"Weaknesses: {score.weaknesses}")
```

```python
# Heuristic evaluation (no API call)
evaluator_h = OutputEvaluator(mode='heuristic')
score_h = evaluator_h.evaluate(ctx, result.response)
print(f"Heuristic overall: {score_h.overall:.1%}")
for dim, val in score_h.dimensions.items():
    print(f"  {dim.value}: {val:.0%}")
```

## 19. Framework Integrations [NO API]

Six framework helpers for seamless integration.

```python
from mycontext.integrations import auto_integrate

ctx = Context(
    guidance=Guidance(role="Expert Analyst", rules=["Be thorough"]),
    directive=Directive(content="Analyze market trends")
)

frameworks = ["langchain", "llamaindex", "crewai", "autogen", "dspy", "semantic_kernel"]
for fw in frameworks:
    try:
        result = auto_integrate(ctx, fw)
        rtype = type(result).__name__
        if isinstance(result, list):
            print(f"{fw:18s} -> {rtype} ({len(result)} items)")
        elif isinstance(result, dict):
            print(f"{fw:18s} -> {rtype} keys={list(result.keys())}")
        elif isinstance(result, str):
            print(f"{fw:18s} -> str ({len(result)} chars)")
        else:
            print(f"{fw:18s} -> {rtype}")
    except ImportError as e:
        print(f"{fw:18s} -> [skip] {e}")
    except Exception as e:
        print(f"{fw:18s} -> [error] {e}")
```

## 21. Three-Tier Execution Model — Summary

| Tier | Function | LLM Calls | Quality | Use Case |
|------|----------|-----------|---------|----------|
| **Static Generic** | `smart_generic_prompt()` | 1 (assess) + 0 (compile) | ~95% | Fast, cost-efficient, chaining |
| **Dynamic Compiled** | `smart_prompt()` | 1 (assess) + 1-3 (refine) | ~95% | Reusable prompt artifacts |
| **Full Response** | `smart_execute()` | 1 (assess) + 1-2 (execute) | ~96% | Deep analysis, maximum nuance |

All three tiers are validated across 6 sprints of controlled experimentation.

```python
import time

from mycontext.intelligence import smart_execute, smart_generic_prompt, smart_prompt

question = "Our cross-functional team has persistent communication breakdowns. Diagnose and solve."

# Tier 1: Static Generic
t0 = time.time()
tier1 = smart_generic_prompt(question, provider=PROVIDER)
t1 = time.time()
tier1_resp = tier1.execute(provider=PROVIDER)
t1b = time.time()
print(f"Tier 1 (Static Generic): prompt={len(tier1.prompt)} chars, response={len(tier1_resp)} chars")
print(f"  Compile: {t1-t0:.1f}s, Execute: {t1b-t1:.1f}s, Total: {t1b-t0:.1f}s")
print(f"  Templates: {tier1.source_templates}")

# Tier 2: Dynamic Compiled
t0 = time.time()
tier2 = smart_prompt(question, provider=PROVIDER)
t1 = time.time()
tier2_resp = tier2.execute(provider=PROVIDER)
t1b = time.time()
print(f"\nTier 2 (Dynamic Compiled): prompt={len(tier2.prompt)} chars, response={len(tier2_resp)} chars")
print(f"  Compile: {t1-t0:.1f}s, Execute: {t1b-t1:.1f}s, Total: {t1b-t0:.1f}s")
print(f"  Templates: {tier2.source_templates}")

# Tier 3: Full Response
t0 = time.time()
tier3_resp, tier3_meta = smart_execute(question, provider=PROVIDER)
t1 = time.time()
print(f"\nTier 3 (Full Response): response={len(tier3_resp)} chars")
print(f"  Total: {t1-t0:.1f}s")
print(f"  Templates: {tier3_meta['templates_used']}")
```
