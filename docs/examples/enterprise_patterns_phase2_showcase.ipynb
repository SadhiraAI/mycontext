# Part 1: Learning & Knowledge Building (5)

**What**: Patterns for scaffolding, memory, cognitive load, and conceptual change.

**Research**: Vygotsky (1978), Ebbinghaus (1885), Sweller (1988), Posner et al. (1982)

**When to use**: EdTech platforms, corporate training, tutoring systems, skill development

## 1. ScaffoldingFramework

**What it does**: Provide temporary support that fades as learner gains competence.

**When to use**: Complex skill training, tutoring, personalized learning

**Industry examples**:
- **Khan Academy**: Adaptive hints that reduce over time
- **Duolingo**: Scaffolded language lessons (translation → production)
- **Code Academy**: Guided coding with decreasing hints

```python
# Example: Coding bootcamp teaching recursion
scaffold = ScaffoldingFramework()

result = scaffold.execute(
    provider="openai",
    task="Learn to write recursive functions in Python",
    current_skill_level="Can write loops and basic functions, unfamiliar with recursion",
    goal="Write recursive solutions for common problems (factorial, fibonacci, tree traversal)"
)

print(result.response[:800] + "...")
```

## 2. SpacedRepetitionOptimizer

**What it does**: Calculate optimal review timing based on Ebbinghaus forgetting curve.

**When to use**: Flashcard systems, language learning, medical education, certification prep

**Industry examples**:
- **Anki**: Spaced repetition flashcards
- **Memrise**: Language learning with SR algorithm
- **Quizlet**: Study reminders based on forgetting curve

```python
# Example: Medical student studying anatomy
sr_optimizer = SpacedRepetitionOptimizer()

result = sr_optimizer.execute(
    provider="openai",
    learning_material="Anatomy: 200 muscle names, origins, insertions, and actions",
    initial_mastery="Just completed module, scored 85% on immediate test",
    previous_performance="After 1 week: 70% recall"
)

print(result.response[:800] + "...")
```

## 3. ZoneOfProximalDevelopment

**What it does**: Identify tasks at "just right" difficulty (Vygotsky's ZPD).

**When to use**: Adaptive learning, curriculum design, challenge calibration

**Industry examples**:
- **DreamBox**: Math software that adapts to ZPD
- **Duolingo**: Difficulty adjustment based on performance
- **Coursera**: Recommended next courses based on skill level

```python
# Example: Data science bootcamp student progression
zpd = ZoneOfProximalDevelopment()

result = zpd.execute(
    provider="openai",
    learner_current_abilities="Can perform basic data cleaning in pandas, create simple visualizations",
    learning_goal="Build end-to-end ML pipeline with model deployment"
)

print(result.response[:800] + "...")
```

## 4. CognitiveLoadManager

**What it does**: Optimize intrinsic, extraneous, and germane cognitive load (Sweller's CLT).

**When to use**: Instructional design, course creation, learning interface design

**Industry examples**:
- **Coursera**: Course design reviews for cognitive load
- **EdX**: Chunking content to prevent overload
- **LinkedIn Learning**: Progressive disclosure of complexity

```python
# Example: Designing a quantum mechanics module
clm = CognitiveLoadManager()

result = clm.execute(
    provider="openai",
    learning_material="Quantum mechanics: wave-particle duality, uncertainty principle, quantum entanglement",
    learner_background="Undergraduate physics majors, completed classical mechanics"
)

print(result.response[:800] + "...")
```

## 5. ConceptualChangeAnalyzer

**What it does**: Help learners revise misconceptions and build accurate mental models.

**When to use**: Science education, technical training, correcting deeply held misunderstandings

**Industry examples**:
- **PhET Simulations**: Interactive tools for physics misconceptions
- **Khan Academy**: Targeted content for common math misconceptions
- **Medical education**: Clinical reasoning error correction

```python
# Example: Physics student with Newton's laws misconception
cca = ConceptualChangeAnalyzer()

result = cca.execute(
    provider="openai",
    topic="Newton's First Law of Motion (inertia)",
    current_understanding="Objects need a continuous force to keep moving at constant velocity"
)

print(result.response[:800] + "...")
```

# Part 2: Evaluation & Assessment (5)

**What**: Patterns for designing rubrics, formative/summative assessment, and peer/self review.

**Research**: Brookhart (2013), Black & Wiliam (1998), Wiggins & McTighe (2005), Boud (1995)

**When to use**: Educational institutions, certification programs, corporate L&D, assessment platforms

## 6. RubricDesigner

**What it does**: Create clear, objective assessment criteria and scoring rubrics.

**When to use**: Grading assignments, performance evaluation, project assessment

**Industry examples**:
- **Turnitin**: Integrated rubric design for educators
- **Canvas LMS**: Rubric builder for course assessments
- **Corporate HR**: Performance review rubrics

```python
# Example: Research paper rubric design
rubric = RubricDesigner()

result = rubric.execute(
    provider="openai",
    assessment_task="Research paper on climate change impacts (2000 words)",
    learning_objectives="Synthesize multiple sources, develop evidence-based argument, write clearly",
    rubric_type="analytic"
)

print(result.response[:800] + "...")
```

## 7. FormativeAssessmentFramework

**What it does**: Design ongoing assessment FOR learning (not just OF learning).

**When to use**: Classroom teaching, tutoring, adaptive learning platforms

**Industry examples**:
- **Kahoot**: Real-time formative assessment games
- **Socrative**: Exit tickets and quick checks
- **Google Classroom**: Embedded formative quizzes

```python
# Example: Math module formative assessment plan
formative = FormativeAssessmentFramework()

result = formative.execute(
    provider="openai",
    learning_unit="Quadratic equations module (2 weeks)",
    learning_goal="Solve quadratic equations using factoring, completing the square, and quadratic formula"
)

print(result.response[:800] + "...")
```

## 8. SummativeEvaluator

**What it does**: Design final evaluation using backward design principles.

**When to use**: Final exams, certification tests, course-end projects

**Industry examples**:
- **ETS**: Standardized test design (GRE, TOEFL)
- **Coursera**: Course completion assessments
- **Professional certifications**: CPA, PMP, AWS exams

```python
# Example: Data science course final assessment
summative = SummativeEvaluator()

result = summative.execute(
    provider="openai",
    course_title="Introduction to Data Science",
    learning_outcomes="Apply statistical methods to data, create visualizations, build predictive models, communicate findings"
)

print(result.response[:800] + "...")
```

**What it does**: Design structured peer review with training and quality control.

**When to use**: Writing courses, project reviews, code review, presentations

**Industry examples**:
