## 2. Core Concepts

### The Building Blocks

mycontext has four core components:

1. **Directive** - What to do (instructions)
2. **Guidance** - How to behave (role, rules, style)
3. **Constraints** - Boundaries and requirements
4. **Context** - The complete package

Let's explore each one:

````python
from mycontext import Context

# Simple context
context = Context(
    guidance="Expert Python Developer",
    directive="Review this code for security issues"
)

print(context.assemble())
````

````python
from mycontext import Constraints, Context, Directive, Guidance

# 1. Directive - The task
directive = Directive(
    content="Analyze customer churn data and identify key drivers",
    priority=8
)

print("Directive:")
print(f"  Content: {directive.content}")
print(f"  Priority: {directive.priority}")
print()
````

````python
# 2. Guidance - The behavior
guidance = Guidance(
    role="Expert Data Scientist with 10+ years experience",
    rules=[
        "Use statistical rigor",
        "Visualize key findings",
        "Explain methodology clearly"
    ],
    expertise=[
        "Python data science stack",
        "Machine learning algorithms",
        "Business intelligence"
    ],
    style="analytical, clear, actionable"
)

print("Guidance:")
print(f"  Role: {guidance.role}")
print(f"  Rules: {len(guidance.rules)} rules defined")
print(f"  Expertise: {len(guidance.expertise) if guidance.expertise else 0} areas")
print(f"  Style: {guidance.style}")
print()
````

````python
# 3. Constraints - The boundaries
constraints = Constraints(
    must_include=[
        "Statistical significance tests",
        "Actionable recommendations"
    ],
    must_not_include=[
        "Unvalidated assumptions",
        "Raw data dumps"
    ],
    style_guide="Use clear headings and bullet points"
)

print("Constraints:")
print(f"  Must include: {constraints.must_include}")
print(f"  Must NOT include: {constraints.must_not_include}")
print()
````

````python
# 4. Context - Putting it all together
context = Context(
    guidance=guidance,
    directive=directive,
    constraints=constraints
)

print("Complete Context:")
print("=" * 60)
print(context.assemble())
print("=" * 60)
````

## 5. Transformation Engine (Auto Pattern Selection)

The **Transformation Engine** is the brain of mycontext. It automatically:
- Analyzes your input
- Detects input type (question, decision, problem, etc.)
- Assesses complexity
- Recommends the best cognitive patterns
- Creates optimized context

### Basic Usage

````python
from mycontext.intelligence import TransformationEngine

# Create engine
engine = TransformationEngine()

# Analyze input
analysis = engine.analyze_input(
    "Should we migrate our monolithic application to microservices?"
)

print("Input Analysis:")
print("=" * 60)
print(f"Input Type: {analysis.input_type.value}")
print(f"Complexity: {analysis.complexity.value}")
print(f"Domain: {analysis.domain}")
print(f"Key Concepts: {analysis.key_concepts}")
print(f"Ambiguity Level: {analysis.ambiguity_level}")
print("\nRecommended Patterns:")
for pattern in analysis.recommended_patterns:
    print(f"  • {pattern}")
print(f"\nConfidence: {analysis.confidence:.1%}")
print("=" * 60)
````

### Automatic Transformation

````python
from mycontext.intelligence import transform

# Convenience function - one line!
context = transform("What are the trade-offs between React and Vue.js?")

print("Auto-Generated Context:")
print("=" * 80)
print(context.assemble())
print("=" * 80)
````

## 6. Quality Metrics

Measure and improve your context quality across 6 dimensions:
1. **Clarity** - How clear and unambiguous
2. **Completeness** - How thorough and comprehensive
3. **Specificity** - How detailed and concrete
4. **Relevance** - How focused on the task
5. **Structure** - How well-organized
6. **Efficiency** - How concise vs verbose

### Evaluating Context Quality

````python
from mycontext import Context, Guidance
from mycontext.intelligence import QualityMetrics

metrics = QualityMetrics()

# Example 1: Basic context
basic_context = Context(guidance="Expert")
basic_score = metrics.evaluate(basic_context)

print("Basic Context Quality:")
print(f"Overall Score: {basic_score.overall:.2f}")
print("\nDimensions:")
for dim, score in basic_score.dimensions.items():
    print(f"  {dim.value}: {score:.2f}")
print("\nSuggestions:")
for suggestion in basic_score.suggestions[:3]:
    print(f"  • {suggestion}")
````

````python
# Example 2: Enhanced context
enhanced_context = Context(
    guidance=Guidance(
        role="Expert Data Engineer with distributed systems expertise",
        rules=[
            "Focus on scalability and performance",
            "Consider cost optimization",
            "Ensure data quality"
        ],
        expertise=["Apache Spark", "Kafka", "AWS"]
    ),
    directive="Optimize data pipeline to handle 10M events/day"
)

enhanced_score = metrics.evaluate(enhanced_context)

print("\nEnhanced Context Quality:")
print(f"Overall Score: {enhanced_score.overall:.2f}")
print("\nDimensions:")
for dim, score in enhanced_score.dimensions.items():
    print(f"  {dim.value}: {score:.2f}")
````

### Quality Comparison

````python
# Compare contexts
comparison = metrics.compare(basic_context, enhanced_context)

print("Quality Comparison:")
print("=" * 60)
print(comparison)
print("=" * 60)
````

### Quality Report

````python
# Generate detailed report
report = metrics.report(enhanced_score)

print(report)
````

## 11. Advanced Techniques

### Technique 1: Multi-Pattern Workflows

### Technique 2: Context Chaining

````python
from mycontext import Constraints, Context, Directive, Guidance

# Start with base context
base_context = Context(
    guidance=Guidance(
        role="Technical Architect",
        rules=["Focus on scalability"]
    )
)

# Build on it - add directive
enhanced_context = Context(
    guidance=base_context.guidance,
    directive=Directive(
        content="Design microservices architecture",
        priority=8
    )
)

# Further enhance - add constraints
final_context = Context(
    guidance=Guidance(
        role=base_context.guidance.role,
        rules=base_context.guidance.rules + [
            "Consider cost optimization",
            "Ensure observability"
        ],
        expertise=["Kubernetes", "Docker", "AWS"]
    ),
    directive=enhanced_context.directive,
    constraints=Constraints(
        must_include=["Service mesh", "API gateway"],
        must_not_include=["Single points of failure"]
    )
)

print("Progressive Context Enhancement:")
print("=" * 80)
print(final_context.assemble())
print("=" * 80)
````

### Technique 3: Dynamic Pattern Selection

````python
from mycontext.intelligence import TransformationEngine
from mycontext.templates.free import (
    DecisionFramework,
    IdeaGenerator,
    ProblemDecomposer,
    QuestionAnalyzer,
)

engine = TransformationEngine()

# Dynamic pattern selection based on input
def smart_context_creation(user_input: str):
    """Automatically select and apply the best pattern."""

    # Analyze input
    analysis = engine.analyze_input(user_input)

    print(f"Input: {user_input}")
    print(f"Detected Type: {analysis.input_type.value}")
    print(f"Recommended Patterns: {analysis.recommended_patterns}")
    print()

    # Map to actual pattern classes
    pattern_map = {
        "question_analyzer": QuestionAnalyzer,
        "decision_framework": DecisionFramework,
        "problem_decomposer": ProblemDecomposer,
        "idea_generator": IdeaGenerator
    }

    # Use first recommended pattern
    if analysis.recommended_patterns:
        pattern_name = analysis.recommended_patterns[0]
        if pattern_name in pattern_map:
            pattern_class = pattern_map[pattern_name]
            print(f"Using pattern: {pattern_name}")
            return True

    return False

# Test with different inputs
test_inputs = [
    "What is the best way to implement caching?",
    "Should we use REST or GraphQL?",
    "The database queries are too slow",
    "Generate ideas for improving user retention"
]

for inp in test_inputs:
    smart_context_creation(inp)
    print("-" * 60)
    print()
````

### Best Practice 1: Start Simple, Iterate

````python
from mycontext import Context
from mycontext.intelligence import QualityMetrics

metrics = QualityMetrics()

# Start simple
simple = Context(
    guidance="Python Expert",
    directive="Optimize code performance"
)

score = metrics.evaluate(simple)
print(f"Initial quality: {score.overall:.2f}")
print("Top suggestions:")
for sug in score.suggestions[:3]:
    print(f"  - {sug}")
print()

# RULE: If quality < 0.7, enhance it
if score.overall < 0.7:
    print("Quality below threshold, enhancing...\n")

    from mycontext import Directive, Guidance
    enhanced = Context(
        guidance=Guidance(
            role="Senior Python Performance Engineer",
            rules=["Profile before optimizing", "Measure impact"]
        ),
        directive=Directive(
            content="Optimize database query performance in user service",
            priority=7
        )
    )

    new_score = metrics.evaluate(enhanced)
    print(f"Enhanced quality: {new_score.overall:.2f}")
    print(f"Improvement: +{new_score.overall - score.overall:.2f}")
````

### Best Practice 3: Always Check Quality

````python
from mycontext import Context
from mycontext.intelligence import QualityMetrics


def create_quality_context(guidance, directive, min_quality=0.7):
    """Helper function that ensures quality threshold."""

    context = Context(guidance=guidance, directive=directive)
    metrics = QualityMetrics()
    score = metrics.evaluate(context)

    print(f"Context Quality: {score.overall:.2f}")

    if score.overall < min_quality:
        print(f"⚠️  Quality below {min_quality}, consider enhancing")
        print("Suggestions:")
        for sug in score.suggestions[:3]:
            print(f"  • {sug}")
    else:
        print(f"✅ Quality meets threshold ({min_quality})")

    return context, score

# Test it
context, score = create_quality_context(
    guidance="Expert Developer",
    directive="Build API",
    min_quality=0.8
)
````
