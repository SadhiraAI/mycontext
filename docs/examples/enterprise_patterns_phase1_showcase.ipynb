## Three Ways to Use Patterns

Every pattern can be used in 3 modes:

### Mode 1: Context-Only (No LLM)

**Use when:**
- Testing/development
- Building contexts for later use
- Exporting to other systems
- No API key available

**Cost:** $0, **Speed:** <1ms

```python
# Example: Build context without calling LLM
pattern = MetacognitiveMonitor()

context = pattern.build_context(
    task_description="Learn Python decorators",
    current_approach="Reading documentation",
    progress_so_far="Understand syntax but confused about closures"
)

# Inspect the context
print("=== GUIDANCE ===")
print(f"Role: {context.guidance.role}")
print("\n=== DIRECTIVE (first 500 chars) ===")
print(context.directive.content[:500])

# Export anywhere
openai_format = context.to_openai()
print(f"\nExported to OpenAI format: {len(openai_format['messages'])} messages")
```

### Mode 2: Full LLM Execution

**Use when:**
- Production applications
- Need actual answers
- Have API key configured

**Cost:** ~$0.01-0.05/call, **Speed:** 2-5s

```python
# Example: Execute with LLM (requires API key)
pattern = MetacognitiveMonitor()

result = pattern.execute(
    provider="gemini",  # or "openai", "anthropic"
    task_description="Learn Python decorators",
    current_approach="Reading documentation",
    progress_so_far="Understand syntax but confused about closures",
    challenges="Don't understand why we need @wraps"
)

print("=== LLM RESPONSE ===")
print(result.response)
print(f"\nCost: ~${result.cost:.4f}")
print(f"Tokens: {result.total_tokens}")
```

### Mode 3: Hybrid (Build + Execute Later)

**Use when:**
- Batch processing
- Need to inspect before executing
- Want to modify context before LLM call
- Building pipelines

**Cost:** Deferred, **Speed:** Flexible

```python
# Example: Build context, inspect, then execute
pattern = MetacognitiveMonitor()

# Step 1: Build context
context = pattern.build_context(
    task_description="Design microservices architecture",
    current_approach="Reading Martin Fowler's articles",
    progress_so_far="Understand benefits but unsure about service boundaries"
)

# Step 2: Inspect/modify if needed
print(f"Context quality score: {context.quality_score:.2f}")
print(f"Directive length: {len(context.directive.content)} chars")

# Step 3: Execute when ready
result = context.execute(provider="openai")
print("\n=== RESPONSE ===")
print(result.response[:500] + "...")
```

# Part 1: Metacognition Patterns (5)

**What is Metacognition?** "Thinking about thinking" - awareness and control of your own cognitive processes.

**Research Foundation:** Flavell (1979), Zimmerman (2002), Schraw & Dennison (1994)

**When to use:**
- Learning systems (EdTech)
- Tutoring applications
- Performance coaching
- Self-improvement tools
- Training programs

## 1. MetacognitiveMonitor

**What it does:** Monitors your thinking process, detects comprehension errors, identifies knowledge gaps.

**When to use:**
- Learning new concepts
- Debugging understanding
- Identifying where you're stuck
- Self-assessment

**Industry Examples:**
- **EdTech:** Khan Academy-style learning assistants that detect where students are confused
- **Corporate Training:** Onboarding systems that identify knowledge gaps
- **Coding Bootcamps:** Help students understand what they don't understand
- **Medical Education:** Monitor resident comprehension during training

```python
# Industry Example: EdTech Learning Assistant
monitor = MetacognitiveMonitor()

# Student learning Python
result = monitor.execute(
    provider="gemini",
    task_description="Understand how Python generators work",
    current_approach="Reading documentation and trying examples",
    progress_so_far="Can write basic generators but confused about yield vs return",
    challenges="Don't understand when generator state is saved vs reset"
)

print("=== METACOGNITIVE ANALYSIS ===")
print(result.response)
```

```python
# Industry Example: Corporate Training (Context-only for batch processing)
monitor = MetacognitiveMonitor()

# Build contexts for 100 employees learning new CRM system
employee_data = [
    {"name": "Alice", "challenge": "Can't find customer history"},
    {"name": "Bob", "challenge": "Confused about lead scoring"},
    # ... 98 more
]

contexts = []
for emp in employee_data[:2]:  # Demo with 2
    context = monitor.build_context(
        task_description="Learn new CRM system",
        current_approach="Watching training videos",
        progress_so_far="Completed basics module",
        challenges=emp["challenge"]
    )
    contexts.append({"employee": emp["name"], "context": context})

print(f"Built {len(contexts)} personalized learning contexts")
print("Ready for batch LLM processing")
```

## 2. SelfRegulationFramework

**What it does:** Cyclical self-regulation: Plan → Monitor → Evaluate → Adjust (Zimmerman's SRL model).

**When to use:**
- Goal setting and tracking
- Project management
- Personal development
- Performance improvement

**Industry Examples:**
- **Fitness Apps:** Peloton-style coaching that adapts based on performance
- **Career Development:** LinkedIn Learning paths that adjust to progress
- **Sales Training:** CRM systems that coach reps through the sales cycle
- **Medical Residents:** Surgery skill development with continuous feedback

```python
# Industry Example: Fitness App Coaching
srl = SelfRegulationFramework()

# FORETHOUGHT: User sets goal
planning = srl.execute(
    provider="gemini",
    goal="Run a 10K in under 50 minutes",
    current_phase="forethought",
    context="Current 10K time: 62 minutes, running 3x/week"
)

print("=== PLANNING PHASE ===")
print(planning.response[:500] + "...")
```

```python
# PERFORMANCE: Mid-training check-in
monitoring = srl.execute(
    provider="gemini",
    goal="Run a 10K in under 50 minutes",
    current_phase="performance",
    context="Week 4 of training",
    current_performance="Current 10K: 57 minutes (-5 min improvement), running 4x/week"
)

print("\n=== MONITORING PHASE ===")
print(monitoring.response[:500] + "...")
```

```python
# REFLECTION: After completing goal
reflection = srl.execute(
    provider="gemini",
    goal="Run a 10K in under 50 minutes",
    current_phase="self-reflection",
    outcome="Achieved 48:30 in 10K race!",
    what_worked="Interval training, consistent schedule",
    what_didnt="Skipped strength training, got minor injury"
)

print("\n=== REFLECTION PHASE ===")
print(reflection.response[:500] + "...")
```

## 5. ErrorDetectionFramework

**What it does:** Systematically identifies mistakes, biases, and reasoning flaws using error taxonomies.

**When to use:**
- Code review
- Quality assurance
- Peer review
- Self-checking work

**Industry Examples:**
- **Software Engineering:** GitHub Copilot-style code review
- **Medical:** Diagnostic error detection systems
- **Finance:** Risk assessment and compliance checking
- **Legal:** Contract review for errors and ambiguities

```python
# Industry Example: Code Review Assistant
detector = ErrorDetectionFramework()

code_to_review = """def quicksort(arr):
    if len(arr) <= 1:
        return arr
    pivot = arr[0]
    left = [x for x in arr if x < pivot]
    right = [x for x in arr if x > pivot]
    return quicksort(left) + [pivot] + quicksort(right)"""

result = detector.execute(
    provider="gemini",
    work_to_check=code_to_review,
    domain="programming",
    error_types_to_check="logic errors, edge cases, performance issues"
)

print("=== ERROR ANALYSIS ===")
print(result.response)
```

# Part 3: Systems Thinking Patterns (6)

**What is Systems Thinking?** Understanding complex systems through feedback loops, leverage points, and emergent behavior.

**Research Foundation:** Meadows (2008), Senge (1990), Forrester (1961)

**When to use:**
- Strategy consulting
- Organizational change
- Policy design
- Business analysis
- Complex problem-solving

## 11. FeedbackLoopIdentifier

**What it does:** Maps reinforcing (R) and balancing (B) feedback loops in systems.

**When to use:**
- Business model analysis
- Growth strategy
- Process improvement
- Understanding system behavior

**Industry Examples:**
- **Consulting (Bain):** Network effects analysis for marketplaces
- **Product (Airbnb):** Supply-demand balance loops
- **Finance:** Debt spirals and credit cycles
- **Healthcare:** Disease progression and treatment response

```python
# Industry Example: Social Media Platform Growth
loops = FeedbackLoopIdentifier()

result = loops.execute(
    provider="gemini",
    system="Social media platform user growth",
    behavior="Explosive growth for 2 years, then plateau, now declining engagement"
)

print("=== FEEDBACK LOOP ANALYSIS ===")
print(result.response)
```
